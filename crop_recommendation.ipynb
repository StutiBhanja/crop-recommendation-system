## Step 1 - Split the data into X (inputs) and y (output)

```
X = df.drop(['label'], axis=1)
```

```
y = df[['label']]
```

## Step 2 - Split the data into train and test (80-20)

```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=2)
```

```
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

## Step 3 - Feature Engineering on Train Data (X_train and y_train)

```
X_train
```

```
# All features are numerical in this dataset
num_df = X_train[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]
```

```
num_df
```

```
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler().set_output(transform='pandas')

num_df_trans = scaler.fit_transform(num_df)
```

```
num_df_trans
```

```
num_df_trans.reset_index(inplace=True)
```

```
X_train_trans = num_df_trans.copy()
X_train_trans.pop('index')
```

```
X_train_trans
```

```
y_train.reset_index(inplace=True)
y_train.pop('index')
```

## Step 4 - Training Model using Algorithm

```
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()

model.fit(X_train_trans, y_train)
```

## Step 5 - Feature Engineering on Test Data

```
X_test
```

```
num_df = X_test[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]
```

```
## num trans
num_df_trans = scaler.transform(num_df)
num_df_trans
```

```
num_df_trans = pd.DataFrame(num_df_trans, columns=['N','P','K','temperature','humidity','ph','rainfall'])
num_df_trans.reset_index(inplace=True)
```

```
X_test_trans = num_df_trans.copy()
X_test_trans.pop('index')
```

```
X_test_trans
```

```
y_test.reset_index(inplace=True)
y_test.pop('index')
```

## Step 6 - Predicting on X_test_trans

```
y_pred = model.predict(X_test_trans)
```

```
y_pred
```

## Step 7 - Passing y_test (actual) and y_pred (model prediction) to evaluate

```
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```
